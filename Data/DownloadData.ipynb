```python
from datasets import load_dataset
import pandas as pd

dataset = load_dataset("sinhala-nlp/SOLD")

for split in dataset.keys():
    df = pd.DataFrame(dataset[split])
    df.to_csv(f"SOLD_{split}.csv", index=False, encoding="utf-8")

print("CSV files saved successfully.")


# Load datasets
train_df = pd.read_csv("SOLD_train.csv")
test_df = pd.read_csv("SOLD_test.csv")

# Combine train and test
combined_df = pd.concat([train_df, test_df], ignore_index=True)

# Keep only text and label
final_df = combined_df.loc[:, ["text", "label"]]

# Optional: convert labels to binary (OFF=1, NOT=0)
final_df["label_binary"] = final_df["label"].map({
    "OFF": 1,
    "NOT": 0
})

# Save to CSV
final_df.to_csv("SOLD_text_label_only.csv", index=False)

print("Dataset created successfully!")
print(final_df.head())

```
