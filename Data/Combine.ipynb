```python
import pandas as pd

# Load datasets
en = pd.read_csv("English.csv")                 # columns: text, label (0/1), label-word
si = pd.read_csv("Sinhala.csv")                 # columns: PhraseNo, text, label ("YES"/"NO")
si_2 = pd.read_csv("SOLD_train.csv")  # columns: text, label ("OFF"/"NOT")

# ---------- English ----------
en_std = en.rename(columns={"label": "label_num"})[["text", "label_num"]].copy()
en_std["lang"] = "en"

# ---------- Sinhala (custom dataset) ----------
si_map = {"YES": 1, "NO": 0, "Yes": 1, "No": 0}
si_std = si.rename(columns={"label": "label_word"})[["text", "label_word"]].copy()
si_std["label_num"] = si_std["label_word"].map(si_map)
si_std["lang"] = "si"
si_std = si_std.drop(columns=["label_word"])

# ---------- Sinhala (SOLD dataset) ----------
sold_map = {"OFF": 1, "NOT": 0}
si2_std = si_2[["text", "label"]].copy()
si2_std["label_num"] = si2_std["label"].map(sold_map)
si2_std["lang"] = "si"
si2_std = si2_std.drop(columns=["label"])

# ---------- Combine all datasets ----------
merged = pd.concat(
    [
        en_std[["text", "label_num", "lang"]],
        si_std[["text", "label_num", "lang"]],
        si2_std[["text", "label_num", "lang"]],
    ],
    ignore_index=True
).rename(columns={"label_num": "label"}) \
 .dropna(subset=["text", "label"])

# Save merged dataset
merged.to_csv("merged_all.csv", index=False)

print("Final merged dataset created")
print(merged.head())
print(merged["lang"].value_counts())
print(merged["label"].value_counts())
```
