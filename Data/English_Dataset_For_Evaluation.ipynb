```python
import pandas as pd

# Load datasets
mlma = pd.read_csv("EN_malma.csv")
conan = pd.read_csv("EN_CONAN.csv")

# -----------------------------
# Process MLMA
# -----------------------------
mlma['label'] = mlma['sentiment'].apply(
    lambda x: 0 if str(x).lower() == 'normal' else 1
)

mlma_final = mlma[['tweet', 'label']].copy()
mlma_final.rename(columns={'tweet': 'text'}, inplace=True)
mlma_final['lang'] = 'en'

# -----------------------------
# Process CONAN
# -----------------------------
conan_final = conan[['hateSpeech']].copy()
conan_final.rename(columns={'hateSpeech': 'text'}, inplace=True)

conan_final['label'] = 1
conan_final['lang'] = 'en'

# -----------------------------
# Merge
# -----------------------------
final_dataset = pd.concat([mlma_final, conan_final], ignore_index=True)

# Remove duplicates
final_dataset = final_dataset.drop_duplicates()

# Save
final_dataset.to_csv("final_english_dataset_evaluation.csv", index=False)

print(final_dataset.head())
print("Total Samples:", len(final_dataset))
```
