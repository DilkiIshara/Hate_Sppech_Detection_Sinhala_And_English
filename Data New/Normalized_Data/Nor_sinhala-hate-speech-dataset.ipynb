```python
import pandas as pd

# Load dataset
si_2 = pd.read_csv("../sinhala-hate-speech-dataset.csv")

# ---------- Normalize Dataset ----------
si2_std = si_2[["comment", "label"]].copy()

# Rename column
si2_std = si2_std.rename(columns={"comment": "text"})

# Ensure labels are integers
si2_std["label"] = si2_std["label"].astype(int)

# Add language column
si2_std["lang"] = "si"

# Remove rows with missing values
si2_std = si2_std.dropna(subset=["text", "label"])

# Optional: remove duplicate texts
si2_std = si2_std.drop_duplicates(subset=["text"])

# Save normalized dataset
si2_std.to_csv(
    "sinhala-hate-speech-dataset_normalized.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Dataset normalized successfully!")
print(si2_std.head())

print("\nColumns:")
print(si2_std.columns.tolist())

print("\nLanguage distribution:")
print(si2_std["lang"].value_counts())

print("\nLabel distribution:")
print(si2_std["label"].value_counts())
```
