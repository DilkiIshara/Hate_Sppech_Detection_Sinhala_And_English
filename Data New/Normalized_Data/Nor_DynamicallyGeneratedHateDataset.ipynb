```python
import pandas as pd

# Load dataset
en = pd.read_csv("../DynamicallyGeneratedHateDataset.csv")

# ---------- Normalize Dataset ----------

# Keep only required columns
en_std = en[["text", "label"]].copy()

# Convert labels to lowercase
en_std["label"] = en_std["label"].str.lower().str.strip()

# Map labels
label_map = {
    "hate": 1,
    "not-hate": 0,
    "nothate": 0,
    "non-hate": 0,
    "normal": 0
}

en_std["label"] = en_std["label"].map(label_map)

# Remove rows with unmapped labels
en_std = en_std.dropna(subset=["label"])

# Convert label to integer
en_std["label"] = en_std["label"].astype(int)

# Add language column
en_std["lang"] = "en"

# Remove missing text
en_std = en_std.dropna(subset=["text"])

# Optional: remove duplicate texts
en_std = en_std.drop_duplicates(subset=["text"])

# Save normalized dataset
en_std.to_csv(
    "DynamicallyGeneratedHateDataset_normalized.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Dataset normalized successfully!")

print(en_std.head())

print("\nColumns:")
print(en_std.columns.tolist())

print("\nLanguage distribution:")
print(en_std["lang"].value_counts())

print("\nLabel distribution:")
print(en_std["label"].value_counts())
```
