```python
import pandas as pd

# Load datasets
si_2 = pd.read_csv("../SOLD_train.csv")  # columns: text, label ("OFF"/"NOT")



# ---------- Sinhala (SOLD dataset) ----------
sold_map = {"OFF": 1, "NOT": 0}
si2_std = si_2[["text", "label"]].copy()
si2_std["label_num"] = si2_std["label"].map(sold_map)
si2_std["lang"] = "si"
si2_std = si2_std.drop(columns=["label"])

# ---------- Combine all datasets ----------
merged = pd.concat(
    [
        si2_std[["text", "label_num", "lang"]],
    ],
    ignore_index=True
).rename(columns={"label_num": "label"}) \
 .dropna(subset=["text", "label"])

# Save merged dataset
merged.to_csv("sold_train_normalized.csv", index=False)

print("Final merged dataset created")
print(merged.head())
print(merged["lang"].value_counts())
print(merged["label"].value_counts())
```
